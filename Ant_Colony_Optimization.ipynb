```python

import random


ROADS = ["North", "South", "East", "West"]


traffic = {
    "North": 35,
    "South": 20,
    "East": 50,
    "West": 15
}


CYCLE_TIME = 60
MIN_GREEN = 10
MAX_GREEN = 30
TIME_STEP = 5

# ACO parameters
NUM_ANTS = 30
ITERATIONS = 100
ALPHA = 1.0
BETA = 2.0
EVAPORATION = 0.3
Q = 100.0


pheromone = {road: 1.0 for road in ROADS}


def calculate_cost(allocation):

    total_vehicles = sum(traffic.values())
    total_cost = 0.0

    for road in ROADS:
        vehicles = traffic[road]
        green = allocation[road]


        capacity = green / 2.0


        remaining_queue = max(0.0, vehicles - capacity)


        total_cost += vehicles / green
        total_cost += 2.0 * remaining_queue


    return total_cost / max(1, total_vehicles)


def construct_solution():

    allocation = {road: MIN_GREEN for road in ROADS}

    remaining = CYCLE_TIME - len(ROADS) * MIN_GREEN
    steps = remaining // TIME_STEP

    for _ in range(steps):
        probabilities = {}
        total = 0.0

        for road in ROADS:
            if allocation[road] + TIME_STEP > MAX_GREEN:
                probabilities[road] = 0.0
                continue


            heuristic = traffic[road] / (allocation[road] + 1)


            score = (
                pheromone[road] ** ALPHA
                * heuristic ** BETA
            )

            probabilities[road] = score
            total += score

        if total == 0:
            break


        pick = random.uniform(0, total)
        cumulative = 0.0

        for road in ROADS:
            cumulative += probabilities[road]

            if pick <= cumulative:
                allocation[road] += TIME_STEP
                break

    return allocation


def update_pheromones(best_allocation, best_cost):

    for road in ROADS:
        pheromone[road] *= (1.0 - EVAPORATION)

    deposit = Q / (best_cost + 1e-9)

    for road in ROADS:
        pheromone[road] += deposit * (
            best_allocation[road] / CYCLE_TIME
        )


def optimize_traffic():
    best_allocation = None
    best_cost = float("inf")

    for iteration in range(ITERATIONS):
        iteration_best = None
        iteration_cost = float("inf")


        for _ in range(NUM_ANTS):
            candidate = construct_solution()


            if sum(candidate.values()) != CYCLE_TIME:
                continue

            cost = calculate_cost(candidate)

            if cost < iteration_cost:
                iteration_cost = cost
                iteration_best = candidate

            if cost < best_cost:
                best_cost = cost
                best_allocation = candidate.copy()

        if iteration_best is not None:
            update_pheromones(iteration_best, iteration_cost)

        if (iteration + 1) % 20 == 0:
            print(
                f"Iteration {iteration + 1}: "
                f"Best cost = {best_cost:.4f}"
            )

    return best_allocation, best_cost


def display_results(allocation, cost):

    print(" OPTIMIZED TRAFFIC SIGNAL PLAN")


    for road in ROADS:
        green = allocation[road]
        vehicles = traffic[road]

        print(
            f"{road:8} | Vehicles: {vehicles:3} | "
            f"Green: {green:2} sec | "
            f"Red: {CYCLE_TIME - green:2} sec"
        )


    print(f"Total green time: {sum(allocation.values())} sec")
    print(f"Estimated congestion cost: {cost:.4f}")



def main():
    print("ANT COLONY TRAFFIC MANAGEMENT SYSTEM")


    print("\nCurrent traffic density:")
    for road in ROADS:
        print(f"{road}: {traffic[road]} vehicles")

    best_allocation, best_cost = optimize_traffic()

    if best_allocation is not None:
        display_results(best_allocation, best_cost)
    else:
        print("No valid signal allocation found.")


if __name__ == "__main__":
    main()
```

```python

import random

ROADS = ["North", "South", "East", "West"]

# Take traffic input from the user
def get_traffic_input():
    traffic = {}

    print("=== Traffic Management System Using ACO ===")
    print("Enter the number of waiting vehicles on each road.\n")

    for road in ROADS:
        while True:
            try:
                vehicles = int(input(f"Enter vehicles on {road} road: "))

                if vehicles < 0:
                    print("Please enter a non-negative number.")
                    continue

                traffic[road] = vehicles
                break

            except ValueError:
                print("Invalid input! Please enter a whole number.")

    return traffic


CYCLE_TIME = 60
MIN_GREEN = 10
MAX_GREEN = 30
TIME_STEP = 5

NUM_ANTS = 30
ITERATIONS = 100
ALPHA = 1.0
BETA = 2.0
EVAPORATION = 0.3
Q = 100.0

pheromone = {road: 1.0 for road in ROADS}

# Global traffic values are initialized at runtime
traffic = {}


def calculate_cost(allocation):
    total_vehicles = sum(traffic.values())
    total_cost = 0.0

    for road in ROADS:
        vehicles = traffic[road]
        green = allocation[road]

        capacity = green / 2.0
        remaining_queue = max(0.0, vehicles - capacity)

        total_cost += vehicles / green
        total_cost += 2.0 * remaining_queue

    return total_cost / max(1, total_vehicles)


def construct_solution():
    allocation = {road: MIN_GREEN for road in ROADS}

    remaining = CYCLE_TIME - len(ROADS) * MIN_GREEN
    steps = remaining // TIME_STEP

    for _ in range(steps):
        probabilities = {}
        total = 0.0

        for road in ROADS:
            if allocation[road] + TIME_STEP > MAX_GREEN:
                probabilities[road] = 0.0
                continue

            heuristic = (traffic[road] + 1) / (
                allocation[road] + 1
            )

            score = (
                pheromone[road] ** ALPHA
                * heuristic ** BETA
            )

            probabilities[road] = score
            total += score

        if total == 0:
            break

        pick = random.uniform(0, total)
        cumulative = 0.0

        for road in ROADS:
            cumulative += probabilities[road]

            if pick <= cumulative:
                allocation[road] += TIME_STEP
                break

    return allocation


def update_pheromones(best_allocation, best_cost):
    for road in ROADS:
        pheromone[road] *= (1 - EVAPORATION)

    deposit = Q / (best_cost + 1e-9)

    for road in ROADS:
        pheromone[road] += (
            deposit * best_allocation[road] / CYCLE_TIME
        )


def optimize_traffic():
    best_allocation = None
    best_cost = float("inf")

    for iteration in range(ITERATIONS):
        iteration_best = None
        iteration_cost = float("inf")

        for _ in range(NUM_ANTS):
            candidate = construct_solution()

            if sum(candidate.values()) != CYCLE_TIME:
                continue

            cost = calculate_cost(candidate)

            if cost < iteration_cost:
                iteration_cost = cost
                iteration_best = candidate.copy()

            if cost < best_cost:
                best_cost = cost
                best_allocation = candidate.copy()

        if iteration_best is not None:
            update_pheromones(iteration_best, iteration_cost)

    return best_allocation, best_cost


def display_results(allocation, cost):
    print("\n=== OPTIMIZED TRAFFIC SIGNAL PLAN ===")

    for road in ROADS:
        green = allocation[road]
        red = CYCLE_TIME - green

        print(
            f"{road} Road: {traffic[road]} vehicles | "
            f"Green: {green} seconds | "
            f"Red: {red} seconds"
        )

    print(f"\nTotal green time: {sum(allocation.values())} seconds")
    print(f"Estimated congestion cost: {cost:.4f}")


def main():
    global traffic

    traffic = get_traffic_input()

    print("\nInput traffic data:")
    for road, vehicles in traffic.items():
        print(f"{road}: {vehicles} vehicles")

    best_allocation, best_cost = optimize_traffic()

    if best_allocation is not None:
        display_results(best_allocation, best_cost)
    else:
        print("No valid traffic signal allocation found.")


if __name__ == "__main__":
    main()
```
