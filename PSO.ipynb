```python
import random


n = int(input("Enter number of bus stops: "))


print("Enter distance matrix:")
distance = []

for i in range(n):
    distance.append(list(map(int, input().split())))



def fitness(route):
    total = 0

    for i in range(n - 1):
        total += distance[route[i]][route[i + 1]]

    return total



particles = []

for i in range(5):
    route = list(range(n))
    random.shuffle(route)
    particles.append(route)



pbest = [route.copy() for route in particles]


gbest = min(pbest, key=fitness).copy()



for iteration in range(50):

    for i in range(5):


        new_route = particles[i].copy()


        if random.random() < 0.7:

            a = random.randint(0, n - 1)
            b = random.randint(0, n - 1)


            stop = gbest[a]

            if stop in new_route:
                index = new_route.index(stop)

                new_route[index], new_route[b] = \
                    new_route[b], new_route[index]


        if random.random() < 0.2:
            a, b = random.sample(range(n), 2)
            new_route[a], new_route[b] = \
                new_route[b], new_route[a]

        particles[i] = new_route


        if fitness(particles[i]) < fitness(pbest[i]):
            pbest[i] = particles[i].copy()


        if fitness(pbest[i]) < fitness(gbest):
            gbest = pbest[i].copy()


print("Best Bus Route:", [x + 1 for x in gbest])
print("Minimum Distance:", fitness(gbest), "km")
```
